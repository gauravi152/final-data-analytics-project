```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported Successfully!")
```

```python
customers = pd.read_csv("olist_customers_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
category_translation = pd.read_csv("product_category_name_translation.csv")
```

```python
print("Customers:", customers.shape)
print("Orders:", orders.shape)
print("Order Items:", order_items.shape)
print("Payments:", payments.shape)
print("Reviews:", reviews.shape)
print("Products:", products.shape)
print("Sellers:", sellers.shape)
print("Category Translation:", category_translation.shape)
```

```python
print("Customers Columns:",customers.columns.tolist())
print("\nOrders Columns:",orders.columns.tolist())
print("\nOrder Items Columns:",order_items.columns.tolist())
print("\nPayments Columns:",payments.columns.tolist())
print("\nReviews Columns:",reviews.columns.tolist())
print("\nProducts Columns:",products.columns.tolist())
print("\nSellers Columns:",sellers.columns.tolist())
```

```python
tables = {
    "Customers":customers,
    "Orders": orders,
    "Order Items":order_items,
    "Payemnts":payments,
    "Reviews":reviews,
    "Products":products,
    "Sellers":sellers,
    "Category Transalation":category_translation
}

for name,df in tables.items():
    print(f"\n{name}")
    print("-"*len(name))
    print(df.dtypes)
```

```python
for name, df in tables.items():
    print(f"\n{name}")
    print("-" * len(name))
    print(df.isnull().sum())
```

```python
print("Orders with missing carrier date:")
print(
    orders[orders["order_delivered_carrier_date"].isnull()]
    ["order_status"]
    .value_counts()
)

print("\nOrders with missing customer delivery date:")
print(
    orders[orders["order_delivered_customer_date"].isnull()]
    ["order_status"]
    .value_counts()
)
```

```python
print("Review Score Missing Values:")
print(reviews["review_score"].isnull().sum())

print("\nReview Score Distribution:")
print(reviews["review_score"].value_counts().sort_index())
```

```python
print("Missing values in Products:")
print(products.isnull().sum())

print("\nProducts with missing category:")
print(products[products["product_category_name"].isnull()].head())

print("\nProducts with missing physical attributes:")
print(
    products[
        products[
            ["product_weight_g",
             "product_length_cm",
             "product_height_cm",
             "product_width_cm"]
        ].isnull().any(axis=1)
    ]
)
```

```python
for name ,df in tables.items():
    print(f"{name}:{df.duplicated().sum()}duplicate rows")
```

```python
for name ,df in tables.items():
    print(f"{name}:{df.shape[0]} rows,{df.shape[1]}columns")
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)
orders["order_approved_at"] = pd.to_datetime(
    orders["order_approved_at"]
)
orders["order_delivered_carrier_date"] = pd.to_datetime(
    orders["order_delivered_carrier_date"]
)
orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"]
)
orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"]
)
order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"]
)
reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"]
)
reviews["review_answer_timestamp"] = pd.to_datetime(
    reviews["review_answer_timestamp"]
)
```

```python
print("Orders date columns:")
print(orders.dtypes)

print("\nOrder Items date column:")
print(order_items["shipping_limit_date"].dtype)

print("\nReviews date columns:")
print(reviews.dtypes)
```

```python
reviews["review_comment_title"] = reviews["review_comment_title"].fillna("No Comment")
reviews["review_comment_message"] = reviews["review_comment_message"].fillna("No Comment")
```

```python
products["product_category_name"] = products["product_category_name"].fillna("Unknown")
```

```python
print("Missing values after cleaning:")

print("\nReviews:")
print(reviews[["review_comment_title", "review_comment_message"]].isnull().sum())

print("\nProducts:")
print(products["product_category_name"].isnull().sum())
```

```python
print("FINAL DATASET VALIDATION")
print("=" * 40)

for name, df in tables.items():
    print(f"\n{name}")
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
```
